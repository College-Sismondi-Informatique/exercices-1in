```python
# Méthode 1 

from distributeur import nombre_billets, billet_suivant

somme = 2780
for _ in range(nombre_billets):
    billet = billet_suivant()
    
    n = somme // billet    
    print(n,"billets de", billet)
    
    somme = somme - n * billet

```

```python
# Méthode 2

from distributeur import nombre_billets, billet_suivant

somme = 2780
for _ in range(nombre_billets):
    billet = billet_suivant()
    
    while somme >= billet :
        print(billet)
        somme = somme - billet

```
