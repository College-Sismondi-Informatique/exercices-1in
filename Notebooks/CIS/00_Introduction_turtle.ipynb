# 1. Les instructions de la tortue

## 1.1 Premier dessin

Un programme turtle a toujours la même structure :

- **première ligne** : `from turtle import *` pour importer les instructions de la tortue,
- **au milieu** : les instructions de dessin,
- **dernière ligne** : `done()` pour lancer le dessin.

Au départ, la tortue est au **centre** de l'écran et regarde vers la **droite**.

Exécutez le programme ci-dessous.

````python
from turtle import *

shape('turtle')   # affiche une tortue au lieu d'une flèche
forward(150)      # avance de 150 pixels
left(90)          # tourne à gauche de 90 degrés
forward(100)      # avance de 100 pixels

done()
````

### À retenir

La tortue se déplace avec 4 instructions :

| Instruction | Exemple | Effet |
|:--|:--|:--|
| `forward(d)` | `forward(100)` | avance de `d` pixels |
| `backward(d)` | `backward(100)` | recule de `d` pixels |
| `left(a)` | `left(90)` | tourne à gauche de `a` degrés |
| `right(a)` | `right(90)` | tourne à droite de `a` degrés |

⚠️ `left` et `right` font **tourner la tortue sur place** : elles ne dessinent rien.

# 2. Répéter avec la boucle `for`

Dans le carré de l'exercice 1, on répète **4 fois** les deux mêmes instructions :

```python
forward(100)
left(90)
```

Comme sur AlgoPython, on peut utiliser une boucle `for` pour répéter un bloc d'instructions :

```python
for _ in range(4):
    forward(100)
    left(90)
```

- Le nombre entre les parenthèses de `range()` indique **combien de fois** on répète.
- Ne pas oublier les **deux-points** `:` à la fin de la ligne du `for`.
- Les instructions répétées sont celles qui sont **indentées** (décalées vers la droite de 4 espaces, ou avec la touche <kbd>Tab</kbd>).

````python
from turtle import *

for _ in range(4):
    forward(100)      # indenté : répété 4 fois
    left(90)          # indenté : répété 4 fois

dot(20)               # pas indenté : exécuté une seule fois, après la boucle

done()
````

# 3. Les variables

Une **variable** est un nom qui contient une valeur. On la crée avec le signe `=` :

```python
cote = 100
```

Ensuite, partout où on écrit `cote`, Python le remplace par sa valeur `100`.

## 3.1 Utiliser une variable dans un dessin

Dans le programme ci-dessous, la longueur du côté n'est écrite qu'**une seule fois**. Pour changer la taille du carré, il suffit de modifier **une seule ligne**.

````python
from turtle import *

cote = 100
couleur = 'purple'

color(couleur)
for _ in range(4):
    forward(cote)
    left(90)

done()
````

## 3.2 Modifier une variable dans une boucle

On peut changer la valeur d'une variable pendant le programme. L'instruction

```python
longueur = longueur + 10
```

signifie : *la nouvelle valeur de `longueur` est l'ancienne valeur + 10*.

Si on place cette instruction dans une boucle, la variable change **à chaque répétition**. Cela permet de faire des dessins qui évoluent, comme cette spirale :

````python
from turtle import *

speed(0)
longueur = 10

for _ in range(30):
    forward(longueur)
    left(90)
    longueur = longueur + 10     # chaque côté est plus long que le précédent

done()
````

# 4. Les fonctions

Comme sur AlgoPython, le mot-clé `def` permet de **définir** une nouvelle instruction, appelée **fonction**. On donne un nom à une suite d'instructions, puis on peut l'utiliser autant de fois qu'on veut.

```python
def carre():          # définition : def + nom + () + :
    for _ in range(4):
        forward(50)
        left(90)

carre()               # appel de la fonction
```

<div style='padding:18px; background-color:papayawhip; border-left:6px solid #e07a20'>

**À retenir**

- on **définit** une fonction une seule fois (avec `def`),
- on **appelle** une fonction autant de fois qu'on veut (en écrivant son nom suivi de `()`),
- si on ne l'appelle pas, **il ne se passe rien** !
- les instructions de la fonction sont **indentées**,
- on écrit les noms de fonction **sans accents** : `carre`, `etoile`, `fenetre`.
</div>

````python
from turtle import *

def carre():
    for _ in range(4):
        forward(50)
        left(90)

carre()           # 1er carré
penup()
forward(80)
pendown()
carre()           # 2e carré
penup()
forward(80)
pendown()
carre()           # 3e carré

done()
````

## 4.1 Revenir au point de départ

Une bonne habitude : à la fin d'une fonction, la tortue doit revenir **à sa position de départ** et **dans la même direction**. Ainsi, on sait toujours où elle se trouve quand on enchaîne plusieurs fonctions.

Une fonction peut aussi **appeler une autre fonction**. Ici, `maison()` utilise `carre()` et `toit()`. Ces deux fonctions doivent être définies **avant** d'être utilisées.

````python
from turtle import *

def carre():
    for _ in range(4):
        forward(100)
        left(90)

def toit():
    for _ in range(3):
        forward(100)
        left(120)

def maison():
    carre()
    # aller en haut à gauche du carré
    left(90)
    forward(100)
    right(90)
    toit()
    # revenir au point de départ, dans la même direction
    right(90)
    forward(100)
    left(90)

maison()

done()
````

# 5. Les fonctions avec paramètres

La fonction `carre()` dessine toujours un carré de la **même taille**. Pour pouvoir choisir la taille à chaque appel, on ajoute un **paramètre** entre les parenthèses :

```python
def carre(cote):          # cote est le paramètre
    for _ in range(4):
        forward(cote)
        left(90)

carre(50)                 # ici, cote vaut 50
carre(120)                # ici, cote vaut 120
```

À chaque appel, le paramètre `cote` **reçoit la valeur** écrite entre les parenthèses.

On peut mettre **plusieurs paramètres**, séparés par des virgules :

````python
from turtle import *

def carre(cote, couleur):
    color(couleur)
    for _ in range(4):
        forward(cote)
        left(90)

carre(50, 'red')
carre(100, 'green')
carre(150, 'blue')

done()
````

## 5.1 Colorier une forme

Pour remplir une forme avec une couleur, on utilise :

- `fillcolor(c)` pour choisir la couleur de remplissage,
- `begin_fill()` **avant** de dessiner la forme,
- `end_fill()` **après** avoir dessiné la forme.

````python
from turtle import *

def carre_plein(cote, couleur):
    fillcolor(couleur)
    begin_fill()
    for _ in range(4):
        forward(cote)
        left(90)
    end_fill()

carre_plein(80, 'orange')
forward(100)
carre_plein(50, 'pink')

done()
````
