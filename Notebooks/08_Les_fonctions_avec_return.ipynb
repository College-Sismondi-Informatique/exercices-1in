## 5.2 Problème : `print()` affiche, mais ne donne pas une valeur réutilisable

Observez le programme suivant.


````python
def aire_carre(cote):
    print("L'aire du carré vaut :", cote * cote)

# PROGRAMME PRINCIPAL
aire_carre(5)
aire_carre(12)
````

Cette fonction est utile pour **afficher** le résultat.

Mais si on veut ensuite **réutiliser** l’aire pour faire un autre calcul, ce n’est pas pratique.

Par exemple, si on veut :
- calculer l’aire d’un carré ;
- puis ajouter 10 à cette aire ;
- puis utiliser ce résultat dans un autre calcul ;

il faut que la fonction **renvoie** réellement la valeur calculée.


!!! danger Idée importante

`print()` sert à **afficher** une information à l’écran.

`return` sert à **renvoyer** une valeur au programme.

Ces deux instructions ne font pas la même chose.

## 5.3 Première fonction avec `return`

Voici une nouvelle version de la fonction `aire_carre`.


````python
def aire_carre(cote):
    return cote * cote

# PROGRAMME PRINCIPAL
resultat = aire_carre(5)
print("L'aire du carré vaut :", resultat)
````

### Que se passe-t-il ?

1. On appelle la fonction avec `aire_carre(5)`.
2. Le paramètre `cote` reçoit la valeur `5`.
3. La fonction calcule `5 * 5`.
4. L’instruction `return` renvoie la valeur `25`.
5. Cette valeur est placée dans la variable `resultat`.
6. Le programme principal peut ensuite afficher ou réutiliser cette valeur.


!!! danger À retenir

L'instruction `return` n'affiche pas de valeur mais retourne une valeur (`int`,`float`,`str`,`bool`, ...).

Elle **retourne une valeur** que le programme peut ensuite, par exemple :
- stocker dans une variable ;
- afficher avec `print()` ;
- utiliser dans un autre calcul ;
- transmettre à une autre fonction.

!!!


## 5.4 Différence entre `print()` et `return`

Comparez attentivement les deux programmes suivants.


````python
def double_affiche(nombre):
    print(nombre * 2)

double_affiche(10)
````

````python
def double_renvoie(nombre):
    return nombre * 2

resultat = double_renvoie(10)
print(resultat)
````

Les deux programmes affichent un résultat.

Mais dans le second programme, la valeur renvoyée peut être **réutilisée**.


````python
def double_renvoie(nombre):
    return nombre * 2

resultat = double_renvoie(10)

print("Le résultat vaut :", resultat)
print("Le résultat + 5 vaut :", resultat + 5)
print("Le résultat multiplié par 3 vaut :", resultat * 3)
````

## 5.8 Utiliser `return` avec Turtle

Même avec Turtle, `return` peut être utile.

Par exemple, on peut créer une petite fonction qui calcule la moitié d’une longueur.


````python
from turtle import *

def moitie(nombre):
    return nombre / 2

def carre_centre(cote):
    demi = moitie(cote)

    penup()
    goto(-demi, -demi)
    pendown()

    for _ in range(4):
        forward(cote)
        left(90)

# PROGRAMME PRINCIPAL
carre_centre(100)
done()
````

Dans ce programme :
- `moitie(nombre)` renvoie la moitié du nombre donné ;
- cette valeur est utilisée pour placer le carré au centre ;
- la fonction `carre_centre(cote)` utilise ce calcul pour commencer au bon endroit.

## 5.10  `return` peut être au milieu du code
Dès que Python rencontre `return`, la fonction **s'arrête immédiatement**.

````python
def test(x):
    print('Début')
    return x
    print('Fin')  # Cette instruction ne sera jamais exécuté !!!

print(test(5))
````

## 5.11 Une fonction peut contenir plusieurs `return`
Python exécute seulement le **premier `return` rencontré**.

````python
def signe(x):
    if x >= 0:
        return 'positif'
    return 'négatif'

print(signe(8))
print(signe(-3))
````

⚠️**Explication:** `print('Fin')` n'est jamais exécuté car la fonction s'arrête sur `return`.

## 5.12 Mini-projet — Calculs pour un dessin Turtle

On souhaite dessiner une maison simple.

Pour cela, on va utiliser une fonction qui calcule la moitié d’une largeur.

Cette valeur permet de mieux placer certains éléments du dessin.


````python
from turtle import *

def moitie(nombre):
    return nombre / 2

def rectangle(largeur, hauteur):
    for _ in range(2):
        forward(largeur)
        left(90)
        forward(hauteur)
        left(90)

def maison(largeur, hauteur):
    demi_largeur = moitie(largeur)

    # Corps de la maison
    rectangle(largeur, hauteur)

    # Toit
    penup()
    goto(0, hauteur)
    pendown()
    goto(demi_largeur, hauteur + demi_largeur)
    goto(largeur, hauteur)
    goto(0, hauteur)

# PROGRAMME PRINCIPAL
maison(120, 80)
done()
````
