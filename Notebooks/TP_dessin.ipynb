- **Être capable de collaborer** sur un projet informatique.
  - Pouvoir se répartir le travail, par exemple en identifiant les formes complexes à implémenter sous forme de fonctions Python, et qui peuvent être développées séparément.
- **S'assurer de pouvoir continuer le programme** même si un·e camarade est absent·e durant un cours. Utiliser par exemple un dossier partagé sur le drive.
- **Être capable de déboguer** un programme (par exemple : tester une fonction séparément, utiliser des commentaires pour identifier un problème).
- **Pouvoir décomposer et structurer** un programme en sous-programmes implémentés par des fonctions.
  - Avoir un processus de développement cohérent avec cette approche : décomposer le développement en étapes successives avec une validation pour chaque étape.
  - Chaque forme doit être implémentée par une fonction, avec ou sans paramètres selon le besoin.
    - Le curseur doit toujours revenir à sa position initiale et à son **orientation initiale**.
- **Être capable de documenter** un programme avec des commentaires.
- **Utiliser des boucles** pour factoriser le code, en particulier pour des formes qui se répètent.
  - Utiliser une variable pour créer des variations dans les répétitions.
- **Utiliser une entrée** (`input`) pour avoir un programme **interactif**.
  - Au moins **une valeur numérique**.
- **Être capable d’utiliser** les différentes fonctions du module Python `turtle` pour tracer des figures avec des épaisseurs, couleurs et formes différentes.
- **Être capable de vectoriser** un dessin avec « un algorithme ».

### Éléments pris en compte dans l’évaluation
#### 1) Structure du code
- Le code est structuré de manière correcte et lisible:
  - **1ʳᵉ partie** : import de la librairie `turtle` (et éventuellement de `random`).
  - **2ᵉ partie** : définition des fonctions.
  - **3ᵉ partie** : programme principal (production du dessin en appelant les fonctions définies dans la partie 2).
- Des variables sont utilisées pour les différents paramètres du dessin et les noms des variables/fonctions décrivent correctement ce qu’elles représentent.
- Des commentaires pertinents sont ajoutés afin de décrire le fonctionnement du code (un commentaire d’en-tête, des *docstrings* pour chaque fonction, et des commentaires pour rendre la lecture facile).
- Chaque fonction est commentée en utilisant une *docstring* (voir l’exemple plus bas).
- Le programme utilise les boucles à bon escient (le même code n’est pas répété plusieurs fois).

#### 2) Résultat
- Le programme fonctionne comme attendu (aucune erreur).
- Originalité : le résultat se démarque des exemples donnés, ce qui indique une bonne compréhension des notions étudiées.

#### 3) Modalités du TP
- Chaque version hebdomadaire du TP a été rendue à temps sur Moodle.
- Le fichier est correctement nommé du type : `projet_dessin_prenom1_prenom2_VX.py`.
- Un dossier commun a été créer et partagé entre les membres du groupe


## Grille d’évaluation (Proposition)

> **Principe général**  
> Un programme qui fonctionne mais qui est mal structuré ne peut pas obtenir une bonne note.  
> La qualité du code est évaluée au même niveau que le rendu visuel.

### 1️⃣ Il n'y a pas de commentaire d'en-tête
Le projet exige un commentaire d’en-tête contenant :
- le ou les auteur·ices,
- la date,
- une description du programme.

**Ce que l’on observe :**
- aucune information sur l’auteur ;
- aucune indication sur l’objectif du programme.

*Sans en-tête, le travail est incomplet.*

### 2️⃣Le code n’est pas structuré

**Ce que l’on observe :**
- Le programme est écrit sous forme d’un **bloc unique**, ligne après ligne.
- Il n’y a aucune séparation entre :
  - la préparation du programme,
  - le dessin des différentes formes,
  - le programme principal.

**Pourquoi c’est un problème :**
- Le projet demande explicitement une **structure claire du programme**.
- Un code non structuré est :
  - difficile à lire,
  - difficile à corriger,
  - difficile à modifier.

*Un programme de qualité n’est pas une simple suite d’instructions.*



### 3️⃣Aucune fonction n’est utilisée

**Ce que l’on observe :**
- Il n’y a **aucune fonction** (`def`) dans le programme.
- Toutes les formes sont dessinées directement dans le programme principal.

**Pourquoi c’est un problème :**
- L’énoncé impose :
  - l’utilisation de **fonctions avec et sans paramètres** ;
  - une fonction par élément visuel important.
- Sans fonctions :
  - le code n’est pas réutilisable,
  - il est impossible de tester une partie séparément,
  - le travail en groupe devient très difficile.

*Une forme complexe doit être définie dans une fonction.*

### 4️⃣ Beaucoup de répétitions inutiles

**Ce que l’on observe :**
- De nombreuses séquences de code identiques sont répétées. L'usage d'une boucle `for` serait utile.
- Exemple :
```python
fillcolor("burlywood")
begin_fill()
forward(500)
right(90)
forward(500)
right(90)
end_fill()
```

### 5️⃣ Aucune interaction avec l’utilisateur

Le cahier des charges impose :
- au moins deux questions à l’utilisateur ;
- une entrée libre et une valeur numérique ;
- l’utilisation de if / elif / else.

**Ce que l’on observe :**
- aucun `input()` ;
- aucune condition ;
- l’utilisateur-trice n’a aucun contrôle sur le dessin.

*Le programme ne peut donc pas être qualifié d’interactif.*

### 6️⃣ Il n’y a aucun commentaire explicatif
Un bon programme doit contenir :
- des commentaires pour expliquer les étapes importantes ;
- des docstrings pour expliquer les fonctions.

**Ce que l’on observe :**
- aucun commentaire ;
- personne ne peut comprendre ce que représente chaque partie du dessin ;
- même l’auteur aura de la peine à s’y retrouver plus tard.

*Un code non commenté est considéré comme non terminé.*



??? info C'est quoi cette License "CC-BY-NC-SA"
La ligne `License` renseigne sur la manière dont vous souhaitez partager et protéger votre travail. Par exemple, pour la licence `CC-BY-NC-SA` on a 

- **CC** (*Creative Commons*): Le nom de la licence .
- **BY** (Attribution) : Il est possible d'utiliser ce programme sans droit d'auteur (CopyLeft) mais il faut citer le(s) auteur(s) du programme.
- **NC** (Non-Commercial) : Il est interdit de gagner de l'argent avec ce programme.
- **SA** (Share Alike) : Il est possible de modifier le programme mais la nouvelle version doit garder cette même licence.
???

!!! tip Astuce 3 : Revenir facilement au point de départ 

Après qu'une fonction ait tracé une figure, il faudrait que la *tortue* revienne au point de départ avec l'orientation initiale. Pour ce faire, on peut stoquer dans des variables la position et l'orientation  initiale:

```python
def ma_figure():
    pos_x = xcor() # coordonnée x au départ
    pos_y = ycor() # coordonnée y au départ
    orientation = heading() # l'orientation au départ
    
    # code de la figure...
    
    deplacer(pos_x, pos_y)  # se remettre à la position de départ
    setheading(orientation) # avec la même orientation
```
!!!

#### Exemple 

Testons le programme ci-dessous

````python
from turtle import *

def arbre(hauteur_tronc, couleur_tronc, couleur_feuilles):
    pendown()
    left(90)
    width(hauteur_tronc/6)                # largeur du tronc
    pencolor(couleur_tronc)
    forward(hauteur_tronc)
    dot(hauteur_tronc, couleur_feuilles)  # feuillage simulé par un gros point
    penup()


# Programme principal
arbre(100, 'brown', 'lime')     # dessin du 1er arbre
forward(70)
arbre(90, 'brown', 'green')     # dessin du 2e arbre
forward(70)
arbre(60, 'brown', 'DarkOliveGreen')     # dessin du 3e arbre

done()
````

Comme après avoir dessiné l'arbre, la *tortue* ne revient pas à la position et à l'orientation initiale, il est difficile de dessiner plusieurs arbres les uns après les autres. En faisant, le modification de l'astuce 2 on a

````python
from turtle import *

def deplacer(x,y):
    penup()
    goto(x,y)
    pendown()

def arbre(hauteur_tronc, couleur_tronc, couleur_feuilles):
    pos_x = xcor() # coordonnée x au départ
    pos_y = ycor() # coordonnée y au départ
    orientation = heading() # l'orientation au départ
    
    pendown()
    left(90)
    width(hauteur_tronc/6)                # largeur du tronc
    pencolor(couleur_tronc)
    forward(hauteur_tronc)
    dot(hauteur_tronc, couleur_feuilles)  # feuillage simulé par un gros point
    penup()
    
    deplacer(pos_x, pos_y)  # se remettre à la position de départ
    setheading(orientation) # avec la même orientation
    
    

# Programme principal
arbre(100, 'brown', 'lime')     # dessin du 1er arbre
forward(70)
arbre(90, 'brown', 'green')     # dessin du 2e arbre
forward(70)
arbre(60, 'brown', 'DarkOliveGreen')     # dessin du 3e arbre

done()
````

#### Exemple 2

On peut utiliser la position de départ pour dessiner une figure avec `goto(x,y)` en utilisant les coordonnées relatives au point de départ. Par exemple, la fonction `coeur()` permettra de dessiner des coeurs avec `goto(x,y)` là où la tortue se trouve

````python
from turtle import *

def deplacer(x,y):
    penup()
    goto(x,y)
    pendown()

def coeur():
    pos_x = xcor() # coordonnée x au départ
    pos_y = ycor() # coordonnée y au départ
    
    # dessin do coeur
    goto(pos_x + 100, pos_y + 100)
    goto(pos_x + 100, pos_y + 150)
    goto(pos_x + 50 , pos_y + 200)
    goto(pos_x + 0  , pos_y + 150)
    goto(pos_x - 50 , pos_y + 200)
    goto(pos_x - 100, pos_y + 150)
    goto(pos_x - 100, pos_y + 100)
    
    goto(pos_x,pos_y)
    
coeur()
deplacer(-200,-200)
coeur()

done()
````

### ii. Ajouter des Docstring pour chaque fonction (obligatoire)
Chaque fois qu’une fonction est définie, il est demandé d’ajouter un commentaire de type **docstring** (c’est un commentaire encadré par des triple apostrophes `''' ... '''`) qui indique :
- ce que fait la fonction,
- l'auteur de la fonction,
- si elle a des paramètres,
- quels sont leurs types,
- ce que représentent ces paramètres,
- et un exemple d'utilisation

**Exemple :**

```python
def arbre(h, c1, c2):
    ''' dessine un arbre
    Auteur : Mathieu Schiess
    param(s) :
    h (int) : la hauteur du tronc de l'arbre
    c1 (str) : la couleur (turtle) du tronc
    c2 (str) : la couleur (turtle) du feuillage

    Exemple: arbre(100, "chocolate", "green")
    '''
    pos_x = xcor() # coordonnée x au départ
    pos_y = ycor() # coordonnée y au départ
    orientation = heading() # l'orientation au départ
    
    pendown()
    left(90)
    width(h/6)                # largeur du tronc
    pencolor(c1)
    forward(h)
    dot(h, c2)  # feuillage simulé par un gros point
    penup()
    
    deplacer(pos_x, pos_y)  # se remettre à la position de départ
    setheading(orientation) # avec la même orientation
```

### i. Variables au début du programme :
Pour modifier plus facilement notre programme, on veut **stocker nos constantes dans des variables** dans la partie 1. Par exemple :


```python
LARGEUR = 300
HAUTEUR = 400
COULEUR_FOND = "cyan"

setup(LARGEUR, HAUTEUR)  # fixe la taille de la fenêtre aux nombres de pixels fixés par les variables
bgcolor(COULEUR_FOND)   # fixe la couleur de l’arrière-plan à la valeur entrée dans la variable couleur_fond
```

Cela nous permet d'avoir un programme plus clair, mais également d'utiliser ces valeurs à plusieurs endroits du programme, par exemple pour se rendre en haut à droite du dessins : 


```python
goto(LARGEUR // 2, HAUTEUR // 2)  
```
!!! info Les constantes

- Les constantes sont des variables dont la valeur est fixe tout au long du programme. On donne un valeur au début du programme qui ne changera jamais.
- En python, il est d'usage de nommer les constantes en utilisement seulement des MAJUSCULES.
