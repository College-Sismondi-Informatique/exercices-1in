# Définir - `def`

Dans ce chapitre, nous allons découvrir comment augmenter le vocabulaire de notre langage de programmation en définissant de nouvelles instructions, qu'on appelle aussi **fonction**. Ceci permet de rendre un code plus compact, mais surtout plus lisible. 


````python
from _validation import *
commencer("03_fonction_simple_corrigé")
````


Par exemple, dessiner un rectangle est assez utile, c'est une forme qu'on pourra réutiliser certainement souvent. Il serait pratique de définir un nom pour ces 8 lignes de code. Pouvons-nous définir de nouvelles commandes ?
```Python
forward(160)
left(90)
forward(100)
left(90)
forward(160)
left(90)
forward(100)
left(90)
```
Avec le mot-clé `def`, nous pouvons définir une nouvelle commande que nous pouvons par exemple appeler `rectangle()`.
Cette façon de créer un raccourci est appelée **définir** une fonction.
Le code à exécuter suit l'expression **`def rectangle():`** et est **indenté** (décalé vers la droite).

Ensuite, il suffit d'écrire `rectangle()` pour dessiner un rectangle. On appelle ceci **appeler** une fonction.

<div class="alert alert-block alert-warning"><b>Rappelez-vous ceci :</b> 
    

- on définit une fonction une seule fois avec `def`,
- si on ne l'appelle pas, elle n'est pas exécutée et il ne se passe rien.
- on appelle une fonction autant de fois que l'on veut (zéro, une ou plusieurs fois),


Définir une fonction nous permet de réduire le nombre de lignes de code nécessaires.
Chaque fois que nous utilisons `rectangle()`, au lieu d'écrire 8 lignes, nous écrivons seulement une ligne de code.

⚠️ **Erreur fréquente** : croire qu’écrire seulement `def ...` exécute le code.  

</div>

!!! note &nbsp; Exercice 1: Les trois rectangles

La fonction `rectangle()` est déjà définie. **Ne modifie pas son contenu.**

1. Appelle la fonction pour dessiner un premier rectangle.
2. Tourne à gauche de 90° puis dessine un deuxième rectangle.
3. Tourne à nouveau à gauche de 90° puis dessine un troisième rectangle.

**Vérification :** trouve dans ton code **une seule définition** de `rectangle`, mais **trois appels**.

````python
from turtle import *

# Définition de la fonction - NE MODIFIEZ PAS LA FONCTION !
def rectangle():
    forward(160)
    left(90)
    forward(100)
    left(90)
    forward(160)
    left(90)
    forward(100)
    left(90)

# Appel de la fonction
for _ in range(3):
    rectangle()
    left(90)

done()
````

## Définir une fonction

Le fait de donner un nom à une séquence d'instructions est appelé **définir une fonction**. Une **définition de fonction** comporte :

1. le mot-clé `def` (définir),
1. le nom de la fonction (`batiment/porte`),
1. les parenthèses `()`,
1. le deux-points `:`,
1. un bloc en indentation.

Qu'est-ce qu'un bloc en indentation ?
<br>C'est un bloc de texte qui comporte des espaces vides à gauche. En Python, ces espaces apparaissent en multiples de 4.

L'indentation est très importante en Python. C'est l'indentation qui indique l'étendue des instructions qui font partie de la fonction.

## Indenter avec un raccourci

Comme l'indentation est tellement importante en Python, il en existe un raccourci.
Il faut d'abord sélectionner **les lignes** de code dont vous voulez changer l'indentation.
Ensuite, vous appuyez sur :

- la touche **tab** pour augmenter l'indentation
- la touche **maj+tab** pour diminuer l'indentation



!!! note &nbsp; Exercice 3

Essayez ces raccourcis dans le code ci-dessous. Transformez le code en deux fonctions `batiment()` et `cheminee()`, que vous appelez ensuite.

````python
from turtle import *

def batiment():
    forward(200)
    right(90)
    forward(100)
    right(90)
    forward(200)
    right(90)
    forward(100)
    right(90)

def cheminee():
    forward(30)
    left(90)
    forward(50)
    left(90)
    forward(30)
    left(90)
    forward(50)
    left(90)

batiment()
forward(30)
cheminee()

done()
````

## Maison avec porte

Une fois qu'une fonction est définie, elle peut être utilisée partout, même dans la définition d'une autre fonction.

Ici, nous avons une fonction `porte()`, qui est utilisée à l'intérieur d'une deuxième fonction `maison()`. Pour que ceci soit possible, la définition de porte doit être placée avant la définition de `maison()`.



!!! note &nbsp; Exercice 4

````python
from turtle import *

def porte():
    forward(20)
    left(90)
    forward(40)
    left(90)
    forward(20)
    left(90)
    forward(40)
    left(90)

def maison():
    forward(100)
    left(90)
    forward(60)
    left(45)
    forward(71)
    left(90)
    forward(71)
    left(45)
    forward(60)
    left(90)
    forward(40)
    porte()
    backward(40)

maison()
penup()
forward(140)
pendown()
maison()

done()
````

````python
from turtle import *

def chaise():
    left(90)
    forward(40)
    right(90)
    forward(40)
    right(90)
    forward(40)
    backward(40)
    right(90)
    forward(40)
    right(90)
    forward(40)
    right(90)

def table():
    left(90)
    forward(60)
    right(90)
    forward(100)
    right(90)
    forward(60)
    left(90)

def lit():
    left(90)
    forward(50)
    right(90)
    forward(120)
    right(90)
    forward(50)
    backward(25)
    right(90)
    forward(120)
    right(90)
    forward(15)
    right(90)
    forward(30)
    right(90)
    forward(15)
    left(90)

penup()
goto(-160, 20)
pendown()
chaise()
penup()
goto(-100, 20)
pendown()
table()
penup()
goto(20, 20)
pendown()
chaise()
penup()
goto(-60, -110)
pendown()
lit()

done()
````
