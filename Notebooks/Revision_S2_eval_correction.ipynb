    Exercice 5 — Conditions if / elif / else <br/></h2>

Écrivez un programme qui demande une note sur 6 et affiche :

- `insuffisant` si la note est plus petite que 4
- `suffisant` si la note est comprise entre 4 et 4.9
- `bien` si la note est comprise entre 5 et 5.4
- `très bien` si la note est au moins 5.5


````python
note = float(input("Saisir une note: "))

if 0<= note < 4:
    print("insuffisant")
elif note < 5:
    print("suffisant")
elif note < 5.5:
    print("bien")
elif note <= 6:
    print("très bien")
else:
    print("Votre note n'est pas comprise entre 0 et 6.")
    
````
